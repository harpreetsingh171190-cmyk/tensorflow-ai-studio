```python
from math import e
```

```python
def sigmoid(x):
    return 1/(1+e**-x)
```

```python
sigmoid(.00001)
```

```python
sigmoid(10000)
```

```python
sigmoid(-1)
```

```python
sigmoid(-10)
```
